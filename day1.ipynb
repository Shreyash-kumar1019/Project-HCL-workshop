```python
def calculate_bill(amount, discount=0, tax=5):

  discounted_amount = amount - (amount * discount / 100)
  total = discounted_amount + (discounted_amount * tax / 100)
  return float(total)
print(calculate_bill(2000, discount=10))
```

```python
import numpy as np
temp = np.array([31,35,29,38,40,33,36])
mask = temp > 34
print(mask)
print(temp[temp>34])

#Counting true as 1 false as 0
print("Days above 34 ",(temp>34).sum())

#Combining condiitons
print(temp[(temp>=30) & (temp<= 38)])
print(temp[(temp <30) | (temp >38)])

#np.where(condition , value_if_true, value_else_false)
print(np.where((temp>34),"Hot","PLeasant"))
```
